# Min-Max Scaling (Normalization)

## 1. What is Min-Max Scaling?

Min-Max Scaling is a feature scaling technique used to transform numerical features into a fixed range, usually **0 to 1**.

It preserves the relative relationship between the original values while changing their scale.

### Example

Original values:

| Age |
| --: |
|  20 |
|  30 |
|  40 |
|  50 |
|  60 |

After Min-Max Scaling:

| Age | Scaled Age |
| --: | ---------: |
|  20 |       0.00 |
|  30 |       0.25 |
|  40 |       0.50 |
|  50 |       0.75 |
|  60 |       1.00 |

---

## 2. Why is Min-Max Scaling used?

Different features can have very different ranges.

For example:

* Age → 20–60
* Salary → 20,000–200,000
* Experience → 0–15

Features with different scales can affect some machine learning algorithms.

Min-Max Scaling transforms them into a common range:

* Age → 0–1
* Salary → 0–1
* Experience → 0–1

This makes the features comparable in scale.

---

## 3. Min-Max Scaling Formula

The standard formula is:

$$
X_{scaled} = \frac{X - X_{min}}{X_{max} - X_{min}}
$$

Where:

* `X` = current value
* `X_min` = minimum value of the feature
* `X_max` = maximum value of the feature
* `X_scaled` = scaled value

---

## 4. Numerical Example

Suppose:

```text
Current Age = 40
Minimum Age = 20
Maximum Age = 60
```

Using the formula:

$$
X_{scaled} = \frac{40 - 20}{60 - 20}
$$

$$
= \frac{20}{40}
$$

$$
= 0.5
$$

Therefore:

```text
40 → 0.5
```

This means that 40 is halfway between the minimum value (20) and maximum value (60).

---

## 5. Important Point

Min-Max Scaling does **not** change the actual meaning of the data.

For example:

```text
20 years → 0.00
40 years → 0.50
60 years → 1.00
```

`0.50` does not mean 0.5 years.

It represents the position of 40 relative to the minimum and maximum values.

---

## 6. Advantages

* Converts features to a common scale.
* Usually produces values between 0 and 1.
* Preserves the relative relationships between values.
* Useful for algorithms that are sensitive to feature scale.
* Particularly useful when a bounded range is desired.

---

## 7. Disadvantage

Min-Max Scaling is **sensitive to outliers**.

Because the calculation uses the minimum and maximum values, an extreme value can change the range and compress most other values into a smaller portion of the 0–1 range.

---

## 8. Standardization vs Min-Max Scaling

| Standardization                  | Min-Max Scaling                                |
| -------------------------------- | ---------------------------------------------- |
| Uses mean and standard deviation | Uses minimum and maximum                       |
| Uses `StandardScaler`            | Uses `MinMaxScaler`                            |
| No fixed output range            | Usually produces 0–1                           |
| Can produce negative values      | Usually does not produce negative values       |
| Sensitive to outliers            | Highly sensitive to minimum and maximum values |

---

## 9. Python Implementation

Scikit-learn provides `MinMaxScaler` for Min-Max Scaling.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
```

### Important

`fit()` learns the minimum and maximum values from the data.

`transform()` uses those learned values to scale the data.

`fit_transform()` performs both operations together.

---

## 10. Key Takeaway

**Min-Max Scaling transforms numerical features into a common range, usually 0 to 1, using the minimum and maximum values of the feature.**

<h1>Using the Scikit-learn</h1>

````python
from sklearn.preprocessing import MinMaxScaler
````

````python
scaler = MinMaxScaler()
````

````python
features = df[["age", "salary" , "experience"]]
````

````python
scaler.fit(features)

# fit()
#   ↓
# Find min & max
#   ↓
# Remember them inside scaler
# fit() learns the minimum and maximum values from each column.
````

````python
scaler.data_max_
````

````python
scaler.data_min_
````

````python
sklearn_scaled = scaler.transform(features)
````

````python
sklearn_scaled
````

````python
scaled_df = pd.DataFrame(sklearn_scaled, columns =["age_scaled", "salary_scaled", "experience_scaled"])
scaled_df
````

````python
df[["age_scaled", "salary_scaled", "experience_scaled"]] = scaled_df
````
